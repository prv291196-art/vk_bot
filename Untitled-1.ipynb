```python
"""
ГИБРИДНАЯ СИСТЕМА УПРАВЛЕНИЯ ИЗНОСОМ ИНСТРУМЕНТА
PINN + Продукционные правила для управления режимами резания

Основано на:
- Zhang et al. (2026) - PINN для TWM
- Pogodaev et al. (2021) - Извлечение продукционных правил
"""

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.neural_network import MLPRegressor
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
import warnings
warnings.filterwarnings('ignore')

# Настройка визуализации
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")
RANDOM_SEED = 42
np.random.seed(RANDOM_SEED)
torch.manual_seed(RANDOM_SEED)

print("✅ Библиотеки загружены")
print(f"   PyTorch version: {torch.__version__}")
```

```python
# ============================================================================
# ЗАГРУЗКА И ПЕРВИЧНЫЙ АНАЛИЗ ДАННЫХ
# ============================================================================

# Загрузка данных (укажите ваш путь)
file_path = r"C:\Users\roman\Desktop\freza\mill.csv"
df = pd.read_csv(file_path)

print("="*80)
print("📁 ЗАГРУЗКА ДАННЫХ")
print("="*80)
print(f"Файл: {file_path}")
print(f"Размер: {df.shape[0]} строк, {df.shape[1]} столбцов")
print(f"\nСтолбцы: {df.columns.tolist()}")

print("\n📊 Статистика числовых признаков:")
display(df.describe())

print("\n🔍 Информация о данных:")
print(df.info())

print(f"\n🧹 Пропуски в данных: {df.isnull().sum().sum()}")
```

```python
# ============================================================================
# ПРЕДОБРАБОТКА ДАННЫХ
# ============================================================================

print("="*80)
print("🔧 ПРЕДОБРАБОТКА ДАННЫХ")
print("="*80)

# Удаление строк с пропущенными VB
df_clean = df.dropna(subset=['VB']).copy()
print(f"🧹 После удаления пропусков: {df_clean.shape[0]} строк")

# Сортировка
df_clean = df_clean.sort_values(['case', 'time'])

# Вычисление скорости износа (производная VB по времени)
df_clean['VB_rate'] = df_clean.groupby('case')['VB'].diff() / df_clean.groupby('case')['time'].diff()
df_clean['VB_rate'] = df_clean.groupby('case')['VB_rate'].fillna(method='bfill')
df_clean['VB_rate'] = df_clean['VB_rate'].replace([np.inf, -np.inf], np.nan)
df_clean['VB_rate'] = df_clean.groupby('case')['VB_rate'].fillna(method='bfill')
df_clean['VB_rate'] = df_clean.groupby('case')['VB_rate'].fillna(method='ffill')

print(f"📊 Скорость износа: min={df_clean['VB_rate'].min():.6f}, "
      f"max={df_clean['VB_rate'].max():.6f}")

# Выбор признаков
feature_columns = ['time', 'DOC', 'feed', 'smcAC', 'smcDC', 
                   'vib_table', 'vib_spindle', 'AE_table', 'AE_spindle']
available_features = [col for col in feature_columns if col in df_clean.columns]
print(f"📋 Доступные признаки: {available_features}")

# Создание датасета
X = df_clean[available_features].values
y_VB = df_clean['VB'].values.reshape(-1, 1)
y_rate = df_clean['VB_rate'].values.reshape(-1, 1)

print(f"\n📊 Размерность X: {X.shape}")
print(f"📊 Размерность y_VB: {y_VB.shape}")
print(f"📊 Размерность y_rate: {y_rate.shape}")

# Нормализация
scaler_X = StandardScaler()
X_scaled = scaler_X.fit_transform(X)

scaler_VB = StandardScaler()
y_VB_scaled = scaler_VB.fit_transform(y_VB)

scaler_rate = StandardScaler()
y_rate_scaled = scaler_rate.fit_transform(y_rate)

print("\n✅ Данные нормализованы")

# Разделение на train/test
X_train, X_test, y_train, y_test, y_rate_train, y_rate_test = train_test_split(
    X_scaled, y_VB_scaled, y_rate_scaled, test_size=0.2, random_state=RANDOM_SEED
)

print(f"\n📊 Train: {X_train.shape[0]}, Test: {X_test.shape[0]}")
```

```python
# ============================================================================
# ОПРЕДЕЛЕНИЕ АРХИТЕКТУРЫ PINN
# ============================================================================

print("="*80)
print("🧠 ОПРЕДЕЛЕНИЕ АРХИТЕКТУРЫ PINN")
print("="*80)

class PINN(nn.Module):
    """
    Physics-Informed Neural Network
    
    F_net: предсказывает износ u(t, x)
    G_net: предсказывает скорость износа du/dt (физическое ограничение)
    """
    
    def __init__(self, input_dim, hidden_dim=64):
        super(PINN, self).__init__()
        
        # Сеть F - предсказание износа
        self.F_net = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.Tanh(),
            nn.Linear(hidden_dim, hidden_dim),
            nn.Tanh(),
            nn.Linear(hidden_dim, hidden_dim),
            nn.Tanh(),
            nn.Linear(hidden_dim, 1)
        )
        
        # Сеть G - физический закон (скорость износа)
        # Вход: x + u + u_t + u_x + u_tt
        self.G_net = nn.Sequential(
            nn.Linear(input_dim + 1 + 1 + input_dim + 1, hidden_dim),
            nn.Tanh(),
            nn.Linear(hidden_dim, hidden_dim),
            nn.Tanh(),
            nn.Linear(hidden_dim, hidden_dim),
            nn.Tanh(),
            nn.Linear(hidden_dim, 1)
        )
        
    def forward(self, x):
        x.requires_grad_(True)
        
        # Предсказание износа
        u = self.F_net(x)
        
        # Производная по времени (первый признак - time)
        u_t = torch.autograd.grad(
            u, x, grad_outputs=torch.ones_like(u), 
            create_graph=True, retain_graph=True
        )[0][:, 0:1]
        
        # Производные по всем признакам
        u_x = torch.autograd.grad(
            u, x, grad_outputs=torch.ones_like(u), 
            create_graph=True, retain_graph=True
        )[0]
        
        # Вторая производная по времени
        u_tt = torch.autograd.grad(
            u_t, x, grad_outputs=torch.ones_like(u_t), 
            create_graph=True, retain_graph=True
        )[0][:, 0:1]
        
        # Вход для G_net
        g_input = torch.cat([x, u, u_t, u_x, u_tt], dim=1)
        u_t_pred = self.G_net(g_input)
        
        return u, u_t, u_t_pred

# Инициализация
input_dim = X_train.shape[1]
model = PINN(input_dim=input_dim, hidden_dim=64)

print(f"✅ PINN создана")
print(f"   Входной слой: {input_dim} признаков")
print(f"   Скрытый слой: 64 нейрона")
print(f"   Параметров: {sum(p.numel() for p in model.parameters())}")
```

```python
# ============================================================================
# ОБУЧЕНИЕ PINN
# ============================================================================

print("="*80)
print("🚀 ОБУЧЕНИЕ PINN")
print("="*80)

# Преобразование в тензоры
X_train_t = torch.FloatTensor(X_train)
y_train_t = torch.FloatTensor(y_train)
y_rate_train_t = torch.FloatTensor(y_rate_train)

X_test_t = torch.FloatTensor(X_test)
y_test_t = torch.FloatTensor(y_test)
y_rate_test_t = torch.FloatTensor(y_rate_test)

# DataLoader
train_dataset = TensorDataset(X_train_t, y_train_t, y_rate_train_t)
train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)

# Оптимизатор
optimizer = optim.Adam(model.parameters(), lr=0.001)
scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, patience=50, factor=0.5)

# Функции потерь
def data_loss(u_pred, u_true):
    return nn.MSELoss()(u_pred, u_true)

def physics_loss(u_t_pred, u_t_true):
    return nn.MSELoss()(u_t_pred, u_t_true)

def monotonicity_loss(u_pred):
    diff = u_pred[1:] - u_pred[:-1]
    return torch.mean(torch.relu(-diff))

# Параметры
epochs = 500
alpha = 0.5  # Вес физической потери
beta = 0.1   # Вес монотонности

history = {'loss': [], 'data_loss': [], 'physics_loss': [], 'mono_loss': []}

print("Начало обучения...\n")

for epoch in range(epochs):
    epoch_loss = 0
    epoch_data = 0
    epoch_phys = 0
    epoch_mono = 0
    
    for batch_X, batch_y, batch_y_rate in train_loader:
        optimizer.zero_grad()
        
        u_pred, u_t_pred, u_t_phys = model(batch_X)
        
        loss_data = data_loss(u_pred, batch_y)
        loss_phys = physics_loss(u_t_phys, batch_y_rate)
        loss_mono = monotonicity_loss(u_pred)
        
        loss = loss_data + alpha * loss_phys + beta * loss_mono
        
        loss.backward()
        optimizer.step()
        
        epoch_loss += loss.item()
        epoch_data += loss_data.item()
        epoch_phys += loss_phys.item()
        epoch_mono += loss_mono.item()
    
    history['loss'].append(epoch_loss / len(train_loader))
    history['data_loss'].append(epoch_data / len(train_loader))
    history['physics_loss'].append(epoch_phys / len(train_loader))
    history['mono_loss'].append(epoch_mono / len(train_loader))
    
    scheduler.step(epoch_loss)
    
    if (epoch + 1) % 50 == 0:
        print(f"Epoch {epoch+1:3d}/{epochs} | Loss: {epoch_loss/len(train_loader):.6f} | "
              f"Data: {epoch_data/len(train_loader):.6f} | Phys: {epoch_phys/len(train_loader):.6f}")

print("\n✅ Обучение PINN завершено!")
```

```python
# ============================================================================
# ОЦЕНКА КАЧЕСТВА PINN
# ============================================================================

print("="*80)
print("📊 ОЦЕНКА КАЧЕСТВА PINN")
print("="*80)

# Предсказания
model.eval()
with torch.no_grad():
    u_pred_test, _, _ = model(X_test_t)
    u_pred_test = u_pred_test.numpy()
    u_true_test = y_test

# Обратное масштабирование
u_pred_orig = scaler_VB.inverse_transform(u_pred_test)
u_true_orig = scaler_VB.inverse_transform(u_true_test)

# Метрики
mse = mean_squared_error(u_true_orig, u_pred_orig)
rmse = np.sqrt(mse)
mae = mean_absolute_error(u_true_orig, u_pred_orig)
r2 = r2_score(u_true_orig, u_pred_orig)

print(f"\n📊 Метрики качества на тестовых данных:")
print(f"   MSE:  {mse:.6f}")
print(f"   RMSE: {rmse:.6f}")
print(f"   MAE:  {mae:.6f}")
print(f"   R²:   {r2:.4f}")

# Визуализация
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# График обучения
ax = axes[0]
ax.plot(history['loss'], label='Total Loss', linewidth=2)
ax.plot(history['data_loss'], label='Data Loss', alpha=0.7)
ax.plot(history['physics_loss'], label='Physics Loss', alpha=0.7)
ax.set_xlabel('Epoch')
ax.set_ylabel('Loss')
ax.set_title('Динамика обучения')
ax.legend()
ax.grid(True, alpha=0.3)

# Предсказание vs Истина
ax = axes[1]
ax.scatter(u_true_orig, u_pred_orig, alpha=0.6, edgecolors='k', linewidth=0.5)
ax.plot([u_true_orig.min(), u_true_orig.max()], 
        [u_true_orig.min(), u_true_orig.max()], 'r--', linewidth=2)
ax.set_xlabel('Истинный износ (VB)')
ax.set_ylabel('Предсказанный износ (VB)')
ax.set_title(f'PINN: Предсказание vs Истина (R² = {r2:.4f})')
ax.grid(True, alpha=0.3)

# Ошибки
ax = axes[2]
errors = u_pred_orig.flatten() - u_true_orig.flatten()
ax.hist(errors, bins=30, edgecolor='black', alpha=0.7, color='steelblue')
ax.axvline(0, color='red', linestyle='--', linewidth=2)
ax.set_xlabel('Ошибка предсказания')
ax.set_ylabel('Частота')
ax.set_title(f'Распределение ошибок (MAE = {mae:.4f})')
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
